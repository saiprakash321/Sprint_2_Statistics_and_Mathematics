## 1. T-Test

### Definition

A T-Test is a statistical test used to compare the mean (average) of a group or the means of two groups.

It helps us check whether the difference between the means is statistically significant.

### Assumptions

- Data should be numerical.
- The data should be approximately normally distributed.
- Observations should be independent.
- For an independent two-sample T-Test, the groups should be independent.

### When to Use

A T-Test is used when:

- Comparing the average of two groups.
- Comparing a sample mean with a known value.
- The sample size is relatively small.

### When Not to Use

Do not use a T-Test when:

- The data is strongly non-normal.
- The observations are not independent.
- You need to compare three or more groups.

### Simple Business Example

A company wants to know whether the average sales before and after a new marketing campaign are different.

A T-Test can be used to compare the two averages.

### Python Example

```python
from scipy.stats import ttest_ind

group1 = [20, 22, 21, 19, 23]
group2 = [25, 27, 26, 24, 28]

t_stat, p_value = ttest_ind(group1, group2)

print("T-Statistic:", t_stat)
print("p-value:", p_value)
```

### Interpretation

If the p-value is less than 0.05:

- Reject the Null Hypothesis.
- There is a statistically significant difference between the two groups.

If the p-value is greater than or equal to 0.05:

- Fail to reject the Null Hypothesis.
- There is not enough evidence to say that the two groups are significantly different.

## 2. Z-Test

### Definition

A Z-Test is a statistical test used to compare a sample mean with a population mean or compare the means of two groups.

It is mainly used when the sample size is large or the population standard deviation is known.

### Assumptions

- Data should be numerical.
- Observations should be independent.
- The population should be approximately normally distributed, especially for small samples.
- Population standard deviation should be known for the standard one-sample Z-Test.

### When to Use

A Z-Test is used when:

- The sample size is large.
- The population standard deviation is known.
- We want to compare a sample mean with a known population mean.

### When Not to Use

Do not use a Z-Test when:

- The sample size is small and population standard deviation is unknown.
- The data does not meet the required assumptions.
- You want to compare three or more groups.

### Simple Business Example

A company knows that the average delivery time is 30 minutes.

The company collects a large sample of delivery times and wants to check whether the average delivery time has changed.

A Z-Test can be used for this purpose.

### Python Example

```python
import numpy as np
from scipy.stats import norm

data = [31, 32, 30, 29, 31, 33, 30, 32, 31, 30]

z = (np.mean(data) - 30) / (3 / np.sqrt(len(data)))
p = 2 * (1 - norm.cdf(abs(z)))

print("Z-Score:", z)
print("p-value:", p)
```

### Interpretation

If the p-value is less than 0.05:

- Reject the Null Hypothesis.
- There is a statistically significant difference.

If the p-value is greater than or equal to 0.05:

- Fail to reject the Null Hypothesis.
- There is not enough evidence to say that the sample mean is significantly different from the population mean.

## 3. Chi-Square Test

### Definition

The Chi-Square Test is used to check whether two categorical variables are related.

### Assumptions

- Data should be categorical.
- Observations should be independent.
- Expected frequencies should be large enough.

### When to Use

Use it when you want to check whether two categories are related.

### When Not to Use

Do not use it for numerical data or when the expected frequencies are very small.

### Simple Business Example

A company wants to know whether customer satisfaction is related to the type of product purchased.

The Chi-Square Test can be used to check this relationship.

### Python Example

```python
from scipy.stats import chi2_contingency

data = [[30, 20], [15, 35]]

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-Square:", chi2)
print("p-value:", p)
```

### Interpretation

If the p-value is less than 0.05:

- Reject the Null Hypothesis.
- There is a significant relationship between the variables.

If the p-value is greater than or equal to 0.05:

- Fail to reject the Null Hypothesis.
- There is not enough evidence to say that the variables are related.

## 4. ANOVA

### Definition

ANOVA (Analysis of Variance) is a statistical test used to compare the means of three or more groups.

### Assumptions

- Data should be numerical.
- Groups should be independent.
- Data should be approximately normally distributed.
- Groups should have similar variance.

### When to Use

Use ANOVA when you want to compare the average values of three or more groups.

### When Not to Use

Do not use ANOVA when there are only two groups or when the data does not meet the required assumptions.

### Simple Business Example

A company wants to compare the average sales of three different stores.

ANOVA can be used to check whether there is a significant difference in their average sales.

### Python Example

```python
from scipy.stats import f_oneway

group1 = [20, 22, 21, 19, 23]
group2 = [25, 27, 26, 24, 28]
group3 = [30, 32, 31, 29, 33]

f_stat, p_value = f_oneway(group1, group2, group3)

print("F-Statistic:", f_stat)
print("p-value:", p_value)
```

### Interpretation

If the p-value is less than 0.05:

- Reject the Null Hypothesis.
- At least one group has a significantly different mean.

If the p-value is greater than or equal to 0.05:

- Fail to reject the Null Hypothesis.
- There is not enough evidence to say that the group means are different.

## 5. Mann-Whitney U Test

### Definition

The Mann-Whitney U Test is a statistical test used to compare two independent groups.

It is useful when the data is not normally distributed.

### When to Use

Use it when:

- You have two independent groups.
- The data is numerical or ordinal.
- The data is not normally distributed.

### Simple Business Example

A company wants to compare customer ratings between two different products.

The Mann-Whitney U Test can be used to check whether the ratings are significantly different.

### Python Example

```python
from scipy.stats import mannwhitneyu

group1 = [10, 12, 15, 14, 13]
group2 = [20, 22, 19, 21, 23]

stat, p = mannwhitneyu(group1, group2)

print("U-Statistic:", stat)
print("p-value:", p)
```

### Interpretation

If the p-value is less than 0.05:

- Reject the Null Hypothesis.
- The two groups are significantly different.

If the p-value is greater than or equal to 0.05:

- Fail to reject the Null Hypothesis.
- There is not enough evidence to say that the groups are different.

## 6. Shapiro-Wilk Test

### Definition

The Shapiro-Wilk Test is used to check whether data is approximately normally distributed.

### When to Use

Use it when:

- You want to check normality.
- You need to decide which statistical test to use.
- You are working with numerical data.

### Simple Example

Before using a T-Test or ANOVA, we can check whether the data is approximately normally distributed.

### Python Example

```python
from scipy.stats import shapiro

data = [12, 15, 14, 13, 16, 15, 14]

stat, p = shapiro(data)

print("p-value:", p)
```

### Interpretation

If the p-value is greater than 0.05:

- Data is approximately normally distributed.

If the p-value is less than 0.05:

- Data may not be normally distributed.
