## 1. Introduction to Probability

### Definition

- Probability is a branch of mathematics that measures how likely an event is to occur.
- It is used when the outcome of an experiment or situation is uncertain.
- Probability values range from 0 to 1, where 0 means the event is impossible and 1 means the event is certain.

### Basic Concept

Probability can be calculated using:

**Probability = Number of Favorable Outcomes / Total Number of Possible Outcomes**

For example, when a fair coin is tossed, there are two possible outcomes: Head and Tail. The probability of getting a Head is 1/2 or 0.5.

### Real-Time Example

When checking the weather forecast, we may see a 70% chance of rain. This means rain is considered more likely than not based on the available information.

### AI/ML Example

Probability is widely used in AI and Machine Learning to estimate the likelihood of different outcomes. For example, a spam detection model can predict the probability that an email is spam.

```python
# Probability of getting a Head
favorable = 1
total = 2

probability = favorable / total

print("Probability of Head:", probability)
```

## 2. Events

### Definition

- An event is a specific outcome or a collection of possible outcomes from a random experiment.
- An event can contain one outcome or several outcomes depending on the experiment.

### Basic Concept

For example, when rolling a die, getting a 6 is a single-outcome event. Getting an even number is an event containing three possible outcomes: 2, 4, and 6.

### Real-Time Example

When selecting a student randomly from a class, selecting a student who has scored above 80 marks can be considered an event.

### AI/ML Example

In a Machine Learning classification problem, predicting whether a customer will purchase a product can be treated as an event. The model can estimate the probability of that event occurring.

```python
# Rolling a die
outcomes = [1, 2, 3, 4, 5, 6]

# Event: getting an even number
event = [2, 4, 6]

print("Even outcomes:", event)
print("Probability:", len(event) / len(outcomes))
```

## 3. Sample Space

### Definition

- Sample space is the complete collection of all possible outcomes of a random experiment.
- It is usually represented using the symbol **S**.
- Every possible outcome of the experiment must be included in the sample space.

### Basic Concept

When tossing a coin once:

**S = {Head, Tail}**

When rolling a die:

**S = {1, 2, 3, 4, 5, 6}**

### Real-Time Example

If a traffic signal can show Red, Yellow, or Green, these three possible outcomes form the sample space for the signal state.

### AI/ML Example

In a fraud detection system, the possible outcomes can be represented as:

**S = {Legitimate, Fraudulent}**

The Machine Learning model estimates the probability of each possible outcome.

```python
# Sample space of tossing a coin
sample_space = ["Head", "Tail"]

print("Sample Space:", sample_space)
```

## 4. Independent Events

### Definition

- Independent events are events where the occurrence of one event does not affect the probability of another event.
- The result of one event has no influence on the result of the other event.

### Basic Concept

For two independent events A and B:

**P(A and B) = P(A) × P(B)**

For example, the result of one coin toss does not affect the result of the next coin toss.

### Real-Time Example

If a fair coin is tossed twice, getting Head on the first toss does not change the probability of getting Head on the second toss.

### AI/ML Example

Independent observations are important in many statistical and Machine Learning methods. For example, when individual training examples are treated as independent observations, the outcome of one observation does not directly determine another.

```python
# Two independent coin tosses
p_head = 1 / 2

probability = p_head * p_head

print("Probability of getting two Heads:", probability)
```

## 5. Dependent Events

### Definition

- Dependent events are events where the occurrence of one event affects the probability of another event.
- The result of the first event changes the possible outcomes or probability of the next event.

### Basic Concept

For dependent events, the probability of the second event depends on what happened in the first event.

**P(A and B) = P(A) × P(B | A)**

### Real-Time Example

Suppose a bag contains red and blue balls. If one ball is removed without replacement, the number of balls remaining changes. Therefore, the probability of selecting a particular color on the second draw depends on the first draw.

### AI/ML Example

In customer behavior analysis, a customer's previous activity can influence the probability of their next action. For example, a customer who has already added a product to a shopping cart may have a different probability of purchasing it.

```python
# Drawing two red balls without replacement
red_balls = 3
total_balls = 5

probability = (red_balls / total_balls) * ((red_balls - 1) / (total_balls - 1))

print("Probability of two Red balls:", probability)
```

## 6. Conditional Probability

### Definition

- Conditional probability is the probability of an event occurring when we already know that another event has occurred.
- It allows us to update the probability of an event using additional information.

### Formula

**P(A | B) = P(A and B) / P(B)**

Here, P(A | B) means the probability of event A occurring given that event B has already occurred.

### Simple Example

Suppose the probability of events A and B occurring together is 0.3, while the probability of B is 0.5.

P(A | B) = 0.3 / 0.5 = 0.6

Therefore, the conditional probability is 0.6 or 60%.

### Real-Time Example

Suppose we want to know the probability that a person carries an umbrella given that it is raining. The information that it is raining changes our estimate of whether the person will carry an umbrella.

### AI/ML Example

Conditional probability is an important concept in classification models such as Naive Bayes. For example, it can be used to estimate the probability that an email is spam given the words present in the email.

```python
# P(A and B) / P(B)
probability_a_and_b = 0.3
probability_b = 0.5

conditional_probability = probability_a_and_b / probability_b

print("Conditional Probability:", conditional_probability)
```

## 7. Bayes' Theorem

### Definition

- Bayes' Theorem is a method for calculating the probability of an event using previous knowledge and new evidence.
- It allows us to update an existing probability when new information becomes available.

### Formula

**P(A | B) = [P(B | A) × P(A)] / P(B)**

Where:

- **P(A | B)** = Probability of A given B
- **P(B | A)** = Probability of B given A
- **P(A)** = Prior probability of A
- **P(B)** = Probability of B

### Real-Time Example

A doctor may use a patient's symptoms and medical test results to estimate the probability that the patient has a particular disease. New test information can change the initial probability.

### AI/ML Example

Bayes' Theorem is used in spam classification. A model can use the words and other features in an email as evidence to calculate the probability that the email belongs to the spam category.

```python
# Bayes' Theorem
p_a = 0.01
p_b_given_a = 0.95
p_b = 0.10

result = (p_b_given_a * p_a) / p_b

print("Probability of A given B:", result)
```

## 8. Probability Rules

### Definition

- Probability rules are mathematical rules used to calculate and combine probabilities of different events.
- They help us determine the probability of individual events, combined events, and events that do not occur.

### Important Rules

**Rule of Complement:**

P(A') = 1 - P(A)

**Addition Rule:**

P(A or B) = P(A) + P(B) - P(A and B)

**Multiplication Rule:**

P(A and B) = P(A) × P(B | A)

### Real-Time Example

When rolling a fair die, the probability of getting an even number is:

P(Even) = 3/6 = 0.5

The possible even outcomes are 2, 4, and 6.

### AI/ML Example

Probability rules are used in Machine Learning to combine the likelihood of different outcomes and make predictions based on multiple pieces of information.

```python
p_a = 0.4
p_b = 0.3
p_a_and_b = 0.1

addition = p_a + p_b - p_a_and_b
complement = 1 - p_a

print("P(A or B):", addition)
print("P(Not A):", complement)
```

## 9. Random Variables

### Definition

- A random variable is a variable that assigns a numerical value to the outcome of a random experiment.
- It allows uncertain outcomes to be represented using numbers so that they can be analyzed mathematically.

### Types

There are two common types of random variables:

- **Discrete Random Variable:** Takes countable values, such as the number of customers visiting a store.
- **Continuous Random Variable:** Can take any value within a range, such as height, weight, or temperature.

### Real-Time Example

The number of customers entering a store in one hour is a discrete random variable because it can be counted.

A customer's waiting time can be treated as a continuous random variable because it can take many possible values.

### AI/ML Example

Random variables are used to represent uncertain quantities in Machine Learning, such as customer purchases, model predictions, demand, and other numerical outcomes.

```python
# Random variable for a dice roll
import random

x = random.randint(1, 6)

print("Random Variable X:", x)
```

## 10. Expected Value

### Definition

- Expected value represents the average outcome we would expect over a large number of repeated trials.
- It is calculated by multiplying each possible outcome by its probability and adding the results.

### Formula

**E(X) = Σ [x × P(x)]**

Where:

- **x** = possible outcome
- **P(x)** = probability of that outcome

### Simple Example

For a fair six-sided die:

E(X) = (1 × 1/6) + (2 × 1/6) + (3 × 1/6) + (4 × 1/6) + (5 × 1/6) + (6 × 1/6)

**E(X) = 3.5**

The value 3.5 cannot appear on a single roll, but it represents the average result we would expect over many rolls.

### Real-Time Example

A business can use expected value to estimate the average revenue it may receive from different possible sales outcomes and their probabilities.

### AI/ML Example

Expected value is used in AI and Machine Learning to estimate expected rewards, costs, risks, and outcomes. It is especially useful when a system needs to make decisions under uncertainty.

```python
# Expected value of a fair die
outcomes = [1, 2, 3, 4, 5, 6]

expected_value = sum(outcomes) / len(outcomes)

print("Expected Value:", expected_value)
```

## Advantages of Probability

- Helps us understand uncertainty and possible outcomes.
- Supports better decision-making using available information.
- Helps analyze patterns and relationships in data.
- Provides a foundation for many statistical and Machine Learning methods.

## Limitations of Probability

- Probability gives an estimate and does not guarantee a specific outcome.
- Results depend on the quality and accuracy of the available data.
- Incorrect assumptions can lead to incorrect probability estimates.
- Real-world situations may involve many factors that are difficult to measure.
- Probability models may not always represent complex real-world situations perfectly.
