# Probability Distributions

## What is a Probability Distribution?

A probability distribution shows the possible outcomes of a random experiment and the probability of each outcome.

### Key Points

- It describes possible outcomes and their probabilities.
- The total probability of all outcomes is **1**.
- It helps us understand and analyze uncertain events.

### Types of Probability Distributions

- **Discrete Distribution** – Used for countable outcomes.
- **Continuous Distribution** – Used for measurable values within a range.

### Distributions Covered

- Bernoulli
- Binomial
- Poisson
- Uniform
- Normal
- Exponential

## 1. Bernoulli Distribution

### Definition

- Bernoulli Distribution is used for a single experiment that has only two possible outcomes.

- These outcomes are usually represented as Success/Failure, Yes/No, or 1/0.

### Formula

$$
P(X=x) = p^x(1-p)^{1-x}
$$

Where:

- $x = 1$ represents success.
- $x = 0$ represents failure.
- $p$ is the probability of success.
- $1-p$ is the probability of failure.

### Where it is used

Bernoulli Distribution is used when there are only two possible outcomes from a single trial.

### Real-World Example

A customer either purchases a product or does not purchase it.

- Purchase → Success
- No Purchase → Failure

### AI/ML Example

Bernoulli Distribution can be used in binary classification problems, such as predicting whether an email is spam or not spam.

```python
import numpy as np

# Probability of success
p = 0.6

# Generate 10 Bernoulli outcomes
data = np.random.binomial(1, p, 10)

print("Bernoulli Outcomes:", data)
```

### Python Visualization

The following graph shows the probability of the two possible outcomes in a Bernoulli Distribution.

```python
import matplotlib.pyplot as plt

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities)

plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

plt.xticks([0, 1], ["Failure", "Success"])

plt.show()
```

### Graph Interpretation

The graph shows two possible outcomes: Failure and Success.

Here, the probability of Failure is 0.4 and the probability of Success is 0.6.

## 2. Binomial Distribution

### Definition

- Binomial Distribution is used to find the number of successes in a fixed number of independent trials.

- Each trial has only two possible outcomes: success or failure.

### Formula

$$
P(X=x) = \binom{n}{x}p^x(1-p)^{n-x}
$$

Where:

- $n$ is the total number of trials.
- $x$ is the number of successes.
- $p$ is the probability of success.
- $1-p$ is the probability of failure.

### Where it is used

Binomial Distribution is used when we repeat the same experiment a fixed number of times and count the number of successes.

### Real-World Example

A company contacts 10 customers and wants to know how many customers will purchase a product.

### AI/ML Example

Binomial Distribution can be used to model the number of successful predictions made by a classification model over a fixed number of independent predictions.

```python
import numpy as np

# Number of trials
n = 10

# Probability of success
p = 0.6

# Generate number of successes
data = np.random.binomial(n, p, 10)

print("Number of successes:", data)
```

### Python Visualization

The following graph shows the probability of getting different numbers of successes in 10 trials.

```python
import numpy as np
import matplotlib.pyplot as plt
import math

n = 10
p = 0.6

x = np.arange(0, n + 1)

probabilities = [
    math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in x
]

plt.bar(x, probabilities)

plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.show()
```

### Graph Interpretation

The graph shows the probability of getting different numbers of successes out of 10 trials.

Here, the probability of success in each trial is 0.6.

## 3. Poisson Distribution

### Definition

- Poisson Distribution is used to find the probability of a certain number of events occurring within a fixed interval of time or space.

- The events should occur independently, and the average rate of occurrence should remain constant.

### Formula

$$
P(X=x) = \frac{e^{-\lambda}\lambda^x}{x!}
$$

Where:

- $x$ is the number of events.
- $\lambda$ is the average number of events in the given interval.
- $e$ is approximately 2.718.

### Where it is used

Poisson Distribution is used for counting how many times an event occurs during a fixed period or within a specific area.

### Real-World Example

A customer service center receives an average of 5 calls per hour. Poisson Distribution can be used to estimate the probability of receiving a certain number of calls in one hour.

### AI/ML Example

In AI/ML, Poisson Distribution can be used to model the number of events or occurrences, such as the number of customer requests received by an online system during a fixed time interval.

```python
import numpy as np

# Average number of events
lam = 5

# Generate 10 Poisson values
data = np.random.poisson(lam, 10)

print("Number of events:", data)
```

### Python Visualization

The following graph shows the probability of different numbers of events occurring when the average number of events is 5.

```python
import numpy as np
import matplotlib.pyplot as plt
import math

lam = 5

x = np.arange(0, 15)

probabilities = [
    (np.exp(-lam) * (lam ** k)) / math.factorial(k)
    for k in x
]

plt.bar(x, probabilities)

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")

plt.show()
```

### Graph Interpretation

The graph shows the probability of different numbers of events occurring in a fixed interval.

Here, the average number of events is 5, so the probability is highest around 5 events.

## 4. Uniform Distribution

### Definition

- Uniform Distribution is a probability distribution where all values within a given range have an equal chance of occurring.

### Formula

$$
f(x) = \frac{1}{b-a}
$$

Where:

- $a$ is the lower limit.
- $b$ is the upper limit.

### Where it is used

Uniform Distribution is used when every value within a specific range is equally likely.

### Real-World Example

Selecting a random number between 1 and 10, where each number has an equal chance of being selected.

### AI/ML Example

Uniform Distribution can be used to generate random values for testing, simulation, and initializing parameters in some Machine Learning methods.

```python
import numpy as np

# Generate 10 random values between 1 and 10
data = np.random.uniform(1, 10, 10)

print("Uniformly Distributed Values:")
print(data)
```

### Python Visualization

The following graph shows values generated from a Uniform Distribution between 1 and 10.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(1, 10, 1000)

plt.hist(data, bins=20)

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")

plt.show()
```

### Graph Interpretation

The graph shows that the values are spread relatively evenly across the selected range.

Since the distribution is uniform, no particular value range is strongly concentrated around the center.

## 5. Normal Distribution

### Definition

- Normal Distribution is a continuous probability distribution with a symmetric, bell-shaped curve.

- Most values are concentrated around the mean, with fewer values appearing farther from the mean.

### Formula

$$
f(x) = \frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}
$$

Where:

- $\mu$ is the mean.
- $\sigma$ is the standard deviation.
- $x$ is the value being considered.

### Where it is used

Normal Distribution is commonly used to describe continuous data such as heights, weights, test scores, and measurement values.

### Real-World Example

The heights of people in a large population can often be approximately represented using a Normal Distribution, with most values occurring around the average height.

### AI/ML Example

Normal Distribution is useful in AI/ML for understanding numerical features, analyzing data, and modeling errors or noise in some statistical and Machine Learning methods.

```python
import numpy as np

# Generate normally distributed data
data = np.random.normal(50, 10, 1000)

print("First 10 values:")
print(data[:10])
```

### Python Visualization

The following graph shows a normally distributed dataset with a mean of 50 and a standard deviation of 10.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution")

plt.show()
```

### Graph Interpretation

The graph has a bell-shaped pattern, with most values concentrated around the mean.

Values become less frequent as they move farther away from the center.

## 6. Exponential Distribution

### Definition

- Exponential Distribution is a continuous probability distribution used to describe the waiting time until an event occurs.

### Formula

$$
f(x) = \lambda e^{-\lambda x}
$$

Where:

- $\lambda$ is the rate of occurrence.
- $x$ is the waiting time.
- $e$ is approximately 2.718.

### Where it is used

Exponential Distribution is commonly used to model the time between events.

### Real-World Example

It can be used to represent the waiting time until the next customer arrives at a service center.

### AI/ML Example

In AI/ML, Exponential Distribution can be used to model waiting times and time intervals between events in systems or processes.

```python
import numpy as np

# Rate of occurrence
rate = 0.5

# Generate exponential values
data = np.random.exponential(1 / rate, 10)

print("Waiting Times:")
print(data)
```

### Python Visualization

The following graph shows waiting times generated using an Exponential Distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

rate = 0.5

data = np.random.exponential(1 / rate, 1000)

plt.hist(data, bins=30)

plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")

plt.show()
```

### Graph Interpretation

The graph shows that shorter waiting times occur more frequently, while longer waiting times occur less frequently.

This creates a distribution with a long tail toward higher waiting times.

## Advantages

- Different distributions can be selected for different types of problems.

- Choosing the wrong distribution can lead to incorrect results.
- Probability distributions often require assumptions about the data.
- Real-world data may not perfectly follow a theoretical distribution.
- Some distributions are more suitable for specific types of data.
- Understanding and selecting the correct distribution can require statistical knowledge.
