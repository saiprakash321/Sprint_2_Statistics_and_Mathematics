### 1. Difference between Mean and Median

Mean is the average of all values in a dataset.
Median is the middle value after arranging the data in order.
Mean can be affected by extreme values, while Median is less affected by them.

### 2. What is Variance?

Variance measures how much the values are spread from the mean.
A small variance means the values are close to the mean.
A large variance means the values are more spread out.

### 3. What is Standard Deviation?

Standard Deviation measures the spread of data around the mean.
A small standard deviation means the values are close to the mean.
A large standard deviation means the values are more spread out.

### 4. Explain Bayes' Theorem with an example.

Bayes' Theorem is used to find the probability of an event using previous information.
It helps update the probability when new information is available.
For example, it can be used to find the probability that an email is spam based on its content.

### 5. Difference between Population and Sample

Population means the complete group that we want to study.
Sample means a smaller group selected from the population.
For example, all students in a college are the population, and 100 selected students are the sample.

### 6. Explain Correlation and Covariance.

Correlation shows the strength and direction of the relationship between two variables.
Covariance shows whether two variables increase or decrease together.
Correlation has a value between -1 and +1, making it easier to interpret.

### 7. What is Skewness?

Skewness shows whether data is distributed more toward one side.
Positive skewness means the data has a longer tail on the right side.
Negative skewness means the data has a longer tail on the left side.

### 8. Explain IQR.

IQR stands for Interquartile Range.
It measures the spread of the middle 50% of the data.
The formula is: IQR = Q3 - Q1.

### 9. What is a p-value?

A p-value helps us decide whether there is enough evidence against the Null Hypothesis.
If the p-value is less than 0.05, we commonly reject the Null Hypothesis.
If the p-value is 0.05 or higher, we fail to reject the Null Hypothesis.

### 10. Difference between Type I and Type II Error.

Type I Error occurs when we reject a true Null Hypothesis.
Type II Error occurs when we fail to reject a false Null Hypothesis.
Type I is commonly called a false positive, while Type II is commonly called a false negative.

```python
12. Detect Outliers using the IQR Method.
```

```python
import numpy as np

data = np.array([10, 12, 15, 18, 20, 22, 25, 50])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

outliers = data[(data < Q1 - 1.5 * IQR) | (data > Q3 + 1.5 * IQR)]

print("Outliers:", outliers)
```

13. Calculate Correlation between two variables.

```python
import numpy as np

x = [10, 20, 30, 40, 50]
y = [12, 25, 28, 45, 48]

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)
```
