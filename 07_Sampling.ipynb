## 1. Sampling

### Definition

Sampling is the process of selecting a smaller group from a larger population to study and understand the population.

The larger group is called the **Population**, and the selected smaller group is called the **Sample**.

### Why Is Sampling Used?

Studying an entire population can be difficult, expensive, or time-consuming. Sampling allows us to collect information from a smaller group and use it to understand the larger population.

### Simple Example

Instead of asking every student in a college about their study habits, we can select 100 students and collect their responses.

Here:

- **Population:** All students in the college
- **Sample:** 100 selected students

### Where It Is Used

Sampling is commonly used in:

- Surveys and market research
- Medical studies
- Customer feedback
- Educational research
- Data analysis and Machine Learning

### Important Point

A good sample should represent the population as closely as possible. A poorly selected sample can lead to biased or inaccurate results.

### Python Example

We can use Python to randomly select a sample from a dataset.

```python
import numpy as np

data = np.arange(1, 101)

sample = np.random.choice(data, size=10, replace=False)

print("Sample:", sample)
```

### Explanation

The code creates a population of 100 values and randomly selects 10 values as a sample.

## 2. Random Sampling

### Definition

Random Sampling is a sampling method where every member of the population has an equal chance of being selected.

The selection is made randomly, so personal choice or preference does not influence which members are selected.

### Why Is It Used?

Random Sampling helps reduce selection bias and gives each member of the population a fair chance of being included in the sample.

### Simple Example

Suppose a college has 1,000 students and we want to select 100 students for a survey.

Using Random Sampling, 100 students are selected randomly from the 1,000 students.

### Where It Is Used

Random Sampling is commonly used in:

- Surveys
- Market research
- Research studies
- Data analysis
- Machine Learning

### Important Point

Random Sampling works best when the population is clearly defined and every member can be identified and selected.

### Python Example

Python can be used to randomly select observations from a dataset.

```python
import numpy as np

# Create a population of 100 students
students = np.arange(1, 101)

# Randomly select 10 students
sample = np.random.choice(students, size=10, replace=False)

print("Random Sample:")
print(sample)
```

### Explanation

The code creates 100 student IDs and randomly selects 10 students without selecting the same student more than once.

This demonstrates how Random Sampling can be performed using Python.

## 3. Stratified Sampling

### Definition

Stratified Sampling is a sampling method where the population is divided into smaller groups called **strata** based on a common characteristic.

A sample is then selected from each group.

### Why Is It Used?

Stratified Sampling helps ensure that important groups in the population are represented in the sample.

### Simple Example

Suppose a college has students from three departments:

- Computer Science
- Mechanical
- Civil

Instead of selecting students from the whole college randomly, we can select students separately from each department.

### Where It Is Used

Stratified Sampling is commonly used in:

- Surveys
- Market research
- Educational studies
- Medical research
- Data analysis

### Important Point

The groups should be clearly defined, and the selected sample should represent each group properly.

### Python Example

Python can be used to select samples separately from different groups in a dataset.

```python
import pandas as pd

data = pd.DataFrame({
    "Department": ["CS"] * 5 + ["Mechanical"] * 5 + ["Civil"] * 5,
    "Student": range(1, 16)
})

sample = data.groupby("Department").sample(n=2, random_state=42)

print("Stratified Sample:")
print(sample)
```

### Explanation

The dataset contains students from three different departments.

The code selects 2 students from each department, ensuring that every department is represented in the sample.

## 4. Cluster Sampling

### Definition

Cluster Sampling is a sampling method where the population is divided into groups called **clusters**.

Instead of selecting individual members from every cluster, some clusters are selected randomly and the members from those clusters are studied.

### Why Is It Used?

Cluster Sampling is useful when the population is large or spread across different locations.

### Simple Example

Suppose a school system has 20 schools and we want to survey students.

Instead of selecting students from all 20 schools, we can randomly select 5 schools and survey the students in those selected schools.

### Where It Is Used

Cluster Sampling is commonly used in:

- Surveys
- Education research
- Market research
- Large population studies
- Geographic research

### Important Point

Clusters are usually formed based on location or existing groups, and the selected clusters should represent the larger population.

### Python Example

Python can be used to randomly select clusters from a dataset.

```python
import numpy as np

# Create 10 clusters
clusters = np.arange(1, 11)

# Randomly select 3 clusters
selected_clusters = np.random.choice(
    clusters, size=3, replace=False
)

print("Selected Clusters:")
print(selected_clusters)
```

### Explanation

The code creates 10 clusters and randomly selects 3 clusters for the study.

This demonstrates the basic idea of Cluster Sampling, where complete groups are selected instead of selecting individual members from the entire population.

## 5. Systematic Sampling

### Definition

Systematic Sampling is a sampling method where members are selected at a fixed interval from an ordered population.

### Why Is It Used?

It provides a simple and organized way to select a sample from a large population.

### Simple Example

Suppose there are 100 students and we need to select 10 students.

We can select every 10th student after choosing a starting point.

### Where It Is Used

Systematic Sampling is commonly used in:

- Surveys
- Quality control
- Market research
- Population studies
- Data collection

### Important Point

The population should be arranged in a suitable order, and the sampling interval should be chosen carefully to avoid biased results.

### Python Example

Python can be used to select every fixed number of observations from a dataset.

```python
import numpy as np

# Create a population of 100 students
students = np.arange(1, 101)

# Select every 10th student
sample = students[::10]

print("Systematic Sample:")
print(sample)
```

### Explanation

The code creates 100 student IDs and selects every 10th student.

This demonstrates the basic idea of Systematic Sampling, where observations are selected at a fixed interval.

## 6. Sampling Bias

### Definition

Sampling Bias occurs when the selected sample does not properly represent the population.

This can happen when some members of the population have a higher or lower chance of being selected.

### Why Is It Important?

Sampling Bias can lead to inaccurate results and incorrect conclusions about the population.

### Simple Example

If we want to know the opinions of all college students but survey only final-year students, the results may not represent all students.

### Where It Is Used

Sampling Bias is important in:

- Surveys
- Market research
- Research studies
- Data analysis
- Machine Learning

### Important Point

A representative sample helps reduce Sampling Bias and produces more reliable results.

### Python Example

Python can be used to compare a biased sample with a more representative sample.

```python
import numpy as np

# Population
population = np.arange(1, 101)

# Biased sample: selects only the first 20 members
biased_sample = population[:20]

# Random sample
random_sample = np.random.choice(population, size=20, replace=False)

print("Biased Sample:")
print(biased_sample)

print("\nRandom Sample:")
print(random_sample)
```

### Explanation

The first sample is biased because it only selects members from one part of the population.

The second sample is selected randomly, giving members a more equal chance of being included.

## 7. Sampling Error

### Definition

Sampling Error is the difference between the result obtained from a sample and the actual value of the entire population.

### Why Does It Occur?

Sampling Error occurs because we study only a sample instead of the complete population.

### Simple Example

If the average height of the entire population is 170 cm, but the average height of the selected sample is 168 cm, the difference is Sampling Error.

### Where It Is Used

Sampling Error is considered in:

- Surveys
- Research studies
- Market research
- Data analysis
- Statistical studies

### Important Point

Sampling Error can be reduced by selecting a suitable and sufficiently large sample.

### Python Example

Python can be used to compare a sample mean with the population mean.

```python
import numpy as np

# Create a population
population = np.arange(1, 101)

# Select a random sample
sample = np.random.choice(population, size=20, replace=False)

population_mean = np.mean(population)
sample_mean = np.mean(sample)

sampling_error = sample_mean - population_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
```

### Explanation

The code calculates the mean of the population and the selected sample.

The difference between these two means represents the Sampling Error.

## 8. Sample Size

### Definition

Sample Size is the number of observations or members selected from a population for a study.

### Why Is It Important?

Choosing an appropriate sample size helps us get reliable results without studying the entire population.

A sample that is too small may not represent the population well, while a very large sample may require more time and resources.

### Simple Example

If a college has 5,000 students and we want to conduct a survey, we may select a smaller number of students instead of asking all 5,000.

### Factors Affecting Sample Size

Sample size can depend on:

- Population size
- Required accuracy
- Variability in the data
- Available time and resources

### Where It Is Used

Sample Size is important in:

- Surveys
- Medical studies
- Market research
- Statistical analysis
- Machine Learning

### Python Example

Python can be used to calculate or select a required number of observations from a dataset.

```python
import numpy as np

# Population of 1000 members
population = np.arange(1, 1001)

# Select a sample of 50 members
sample = np.random.choice(population, size=50, replace=False)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
```

### Explanation

The code creates a population of 1,000 members and randomly selects 50 members as the sample.

Here, 1,000 is the population size and 50 is the sample size.
