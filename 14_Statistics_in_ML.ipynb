## Introduction to Statistics in Machine Learning

### Definition

Statistics in Machine Learning helps us understand, analyze, and make decisions from data.

It is used to clean data, prepare features, build models, and evaluate their results.

### Why Is Statistics Important in Machine Learning?

- Helps understand data.
- Helps clean and prepare data.
- Helps find patterns and relationships.
- Helps select useful features.
- Helps evaluate model performance.
- Helps make better predictions.



## 1. Data Cleaning

### Definition

Data Cleaning is the process of finding and fixing incorrect, incomplete, or unnecessary data.

### How Statistics Is Used

Statistics helps us understand the data and find problems such as missing values, incorrect values, and unusual values.

We can use **mean, median, and mode** to understand and clean the data.

### Simple Example

If the age values are:

20, 21, 22, 200, 23

The value **200** may be incorrect or unusual.

### Why Is It Important?

Clean data helps us:

- Build better Machine Learning models.
- Reduce errors.
- Get reliable results.

### Python Example

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 21, None, 23, 24]
})

data["Age"] = data["Age"].fillna(data["Age"].median())

print(data)
```

### Explanation

The dataset has one missing age value.

We use the **median** of the available ages to fill the missing value.

This is a simple example of using statistics for data cleaning.

## 2. Missing Value Handling

### Definition

Missing Value Handling is the process of finding and dealing with empty or missing values in a dataset.

### How Statistics Is Used

Statistics helps us decide how to fill missing values.

Common methods are:

- **Mean** → Average value
- **Median** → Middle value
- **Mode** → Most common value

### Simple Example

If the age values are:

20, 21, 22, 23, missing

We can use the **median** of the available values to fill the missing value.

### Why Is It Important?

Handling missing values helps us:

- Avoid errors during analysis.
- Prepare data for Machine Learning.
- Improve the quality of the dataset.

### Python Example

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 21, None, 23, 24]
})

data["Age"] = data["Age"].fillna(data["Age"].median())

print(data)
```

### Explanation

The dataset contains one missing age value.

The `median()` is used to find the middle value, and `fillna()` replaces the missing value with that median.

This is a simple statistical method for handling missing data.

## 3. Outlier Detection

### Definition

Outlier Detection is the process of finding values that are very different from the other values in a dataset.

### How Statistics Is Used

Statistics helps us identify unusual values using methods such as:

- IQR (Interquartile Range)
- Z-Score
- Mean and Standard Deviation

### Simple Example

If the ages are:

20, 21, 22, 23, 24, 80

The value **80** may be an outlier because it is very different from the other values.

### Why Is It Important?

Outlier detection helps us:

- Find unusual data.
- Reduce errors in analysis.
- Improve data quality.
- Build better Machine Learning models.

### Python Example

```python
import numpy as np

data = [20, 21, 22, 23, 24, 80]

mean = np.mean(data)
std = np.std(data)

print("Mean:", mean)
print("Standard Deviation:", std)
```

### Explanation

The code calculates the **mean** and **standard deviation** of the data.

A value that is very far from the mean may be an outlier.

In this example, **80** is much higher than the other values, so it may be an outlier.

## 4. Feature Engineering

### Definition

Feature Engineering is the process of creating or changing features to make the data more useful for a Machine Learning model.

### How Statistics Is Used

Statistics helps us understand the data and decide which features are useful.

We can use:

- Mean
- Median
- Standard Deviation
- Correlation

to understand and improve features.

### Simple Example

Suppose we have:

- Height
- Weight

We can create a new feature called **BMI** using these values.

This new feature may help the model make better predictions.

### Why Is It Important?

Feature Engineering helps us:

- Create useful features.
- Improve model performance.
- Remove unnecessary information.
- Make data easier for the model to learn.

### Python Example

```python
import pandas as pd

data = pd.DataFrame({
    "Height": [1.6, 1.7, 1.8],
    "Weight": [60, 70, 80]
})

data["BMI"] = data["Weight"] / data["Height"]**2

print(data)
```

### Explanation

The code creates a new **BMI** feature using Height and Weight.

This is an example of Feature Engineering, where existing features are used to create a new useful feature.

## 5. Feature Scaling

### Definition

Feature Scaling is the process of bringing different features to a similar range of values.

### How Statistics Is Used

Statistics helps us understand the range and spread of features.

Common scaling methods include:

- Standardization
- Normalization

### Simple Example

Suppose we have:

- Age: 20–60
- Salary: 20,000–1,00,000

The values are on very different scales.

Feature Scaling brings them to a similar range.

### Why Is It Important?

Feature Scaling helps:

- Prevent large values from dominating.
- Improve the performance of some Machine Learning models.
- Make features easier to compare.

### Python Example

```python
!pip install scikit-learn
```

```python
from sklearn.preprocessing import StandardScaler

data = [[20, 20000], [30, 40000], [40, 60000]]

scaled = StandardScaler().fit_transform(data)

print(scaled)
```

### Explanation

The code uses StandardScaler to scale the features.

The values are changed to a similar scale based on their mean and standard deviation.

This is useful for Machine Learning models that are affected by feature size.

## 6. Data Normalization

### Definition

Data Normalization is the process of changing data to a common range, usually between **0 and 1**.

### How Statistics Is Used

Statistics helps us find the minimum and maximum values and scale the data into a common range.

### Simple Example

Suppose the values are:

10, 20, 30, 40, 50

After normalization, the values will be between **0 and 1**.

### Why Is It Important?

Data Normalization helps to:

- Bring features to a similar range.
- Prevent large values from dominating.
- Improve some Machine Learning models.

### Python Example

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

normalized = (data - data.min()) / (data.max() - data.min())

print(normalized)
```

### Explanation

The code uses the minimum and maximum values to convert the data into a range from **0 to 1**.

The smallest value becomes **0** and the largest value becomes **1**.

## 7. Model Evaluation

### Definition

Model Evaluation is the process of checking how well a Machine Learning model performs.

### How Statistics Is Used

Statistics helps us compare the model's predictions with the actual values.

Different evaluation measures are used for different types of models.

### Simple Example

If a model predicts:

- Actual value = 100
- Predicted value = 90

The difference between them helps us understand how accurate the model is.

### Why Is It Important?

Model Evaluation helps us:

- Check model performance.
- Compare different models.
- Find prediction errors.
- Select a better model.

### Python Example

```python
from sklearn.metrics import mean_absolute_error

actual = [100, 200, 300]
predicted = [90, 210, 280]

error = mean_absolute_error(actual, predicted)

print("Mean Absolute Error:", error)
```

### Explanation

The code calculates the **Mean Absolute Error (MAE)**.

MAE shows the average difference between actual and predicted values.

A **lower MAE** means the predictions are closer to the actual values.

## 8. Linear Regression

### Definition

Linear Regression is a Machine Learning method used to predict a numerical value.

It finds the relationship between an input feature and a target value.

### How Statistics Is Used

Statistics helps Linear Regression understand the relationship between variables and measure how well the model fits the data.

### Simple Example

We can use:

- Study Hours → Input
- Marks → Target

The model can learn how study hours are related to marks and predict marks for new study hours.

### Why Is It Important?

Linear Regression helps us:

- Predict numerical values.
- Understand relationships between variables.
- Make data-based predictions.

### Python Example

```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4]]
y = [45, 50, 60, 65]

model = LinearRegression()
model.fit(X, y)

print(model.predict([[5]]))
```

### Explanation

The model learns the relationship between **Study Hours** and **Marks**.

It then predicts the marks for a student who studies for **5 hours**.

This shows how statistics and Machine Learning can be used together for prediction.

## 9. Logistic Regression

### Definition

Logistic Regression is a Machine Learning method used to predict categories or classes.

It usually gives an output such as **Yes/No** or **0/1**.

### How Statistics Is Used

Statistics helps Logistic Regression understand the relationship between input features and the probability of a particular class.

### Simple Example

We can use:

- Study Hours → Input
- Pass/Fail → Target

The model can predict whether a student is likely to pass or fail.

### Why Is It Important?

Logistic Regression helps us:

- Predict categories.
- Calculate probabilities.
- Make classification decisions.

### Python Example

```python
from sklearn.linear_model import LogisticRegression

X = [[1], [2], [3], [4], [5]]
y = [0, 0, 0, 1, 1]

model = LogisticRegression()
model.fit(X, y)

print(model.predict([[4]]))
```

### Explanation

The model learns the relationship between **Study Hours** and **Pass/Fail**.

Here, `0` represents **Fail** and `1` represents **Pass**.

The model predicts the class for a student who studies for **4 hours**.

## 10. Naive Bayes

### Definition

Naive Bayes is a Machine Learning method used for classification.

It uses probability to predict which category a data point belongs to.

### How Statistics Is Used

Naive Bayes uses probability and statistical calculations to make predictions.

### Simple Example

We can use Naive Bayes to classify an email as:

- Spam
- Not Spam

The model looks at the features of the email and predicts its category.

### Why Is It Important?

Naive Bayes helps us:

- Classify data.
- Make predictions using probability.
- Work well with text data.

### Python Example

```python
from sklearn.naive_bayes import GaussianNB

X = [[1], [2], [3], [4], [5]]
y = [0, 0, 0, 1, 1]

model = GaussianNB()
model.fit(X, y)

print(model.predict([[4]]))
```

### Explanation

The model learns the relationship between the input values and their classes.

It uses probability to predict the class of a new value.

Here, `0` and `1` represent the two different classes.

## 11. Clustering

### Definition

Clustering is a Machine Learning technique used to group similar data points together.

It is an **unsupervised learning** method because the data does not have predefined labels.

### How Statistics Is Used

Statistics helps us understand how similar or different the data points are.

### Simple Example

A company can group customers based on:

- Age
- Spending
- Purchase frequency

The customers with similar behavior can be placed in the same group.

### Why Is It Important?

Clustering helps us:

- Find groups in data.
- Understand customer behavior.
- Discover hidden patterns.
- Segment data into meaningful groups.

### Python Example

```python
from sklearn.cluster import KMeans

data = [[1], [2], [3], [8], [9], [10]]

model = KMeans(n_clusters=2, random_state=0)
model.fit(data)

print(model.labels_)
```

### Explanation

The model divides the data into **2 groups** based on similarity.

Values close to each other are grouped together.

This is a simple example of using Machine Learning to find groups in data.

## 12. Principal Component Analysis (PCA)

### Definition

PCA is a technique used to reduce the number of features in a dataset while keeping the most important information.

### How Statistics Is Used

PCA uses statistical concepts such as **variance** and **correlation** to find the most important directions in the data.

### Simple Example

Suppose a dataset has 10 features.

PCA can reduce these 10 features to 2 or 3 important components while keeping most of the useful information.

### Why Is It Important?

PCA helps us:

- Reduce the number of features.
- Make data easier to visualize.
- Reduce unnecessary information.
- Make some Machine Learning models faster.

### Python Example

```python
from sklearn.decomposition import PCA

data = [[1, 2], [2, 3], [3, 4], [4, 5]]

pca = PCA(n_components=1)
result = pca.fit_transform(data)

print(result)
```

### Explanation

The code reduces the dataset from **2 features to 1 component**.

PCA tries to keep the most important information while reducing the number of features.

## 13. Recommendation Systems

### Definition

A Recommendation System is a Machine Learning system that suggests items to users based on their interests or past behavior.

### How Statistics Is Used

Statistics helps us understand user behavior, preferences, ratings, and similarities between users or items.

### Simple Example

A shopping website may recommend products based on:

- Previous purchases
- Products viewed
- Ratings
- User interests

For example, if a user often buys books, the system may recommend similar books.

### Why Is It Important?

Recommendation Systems help to:

- Give personalized suggestions.
- Improve user experience.
- Help users find useful items.
- Increase engagement.

### Python Example

```python
import numpy as np

ratings = np.array([
    [5, 4, 0],
    [4, 5, 3],
    [0, 4, 5]
])

print("Average ratings:", ratings.mean(axis=0))
```

### Explanation

The code calculates the average rating for each item.

This type of information can help understand which items are more popular among users.

It is a simple example of how statistics can support recommendation systems.
