# Introduction to Statistical Data Visualization

### Definition

Statistical Data Visualization is the process of using graphs and charts to understand data easily.

It helps us see patterns, trends, relationships, distributions, and unusual values in data.

### Why Is It Important?

- Makes data easier to understand.
- Helps identify patterns and trends.
- Helps compare different values.
- Helps find unusual data points.
- Makes data analysis easier.


## 1. Histogram

### Definition

A Histogram is a graph used to represent the distribution of numerical data.

It divides the data into different ranges, called bins, and shows how many values fall into each range.

### Purpose

A Histogram helps us understand the overall pattern of numerical data.

It shows:

- Where most values are located.
- How the data is spread.
- Which ranges have more or fewer values.
- Whether the data is roughly symmetric or skewed.
- Possible unusual values or outliers.

### When to Use

Use it to:

- Understand how data is distributed.
- See where most values are located.
- Identify the spread of data.
- Find possible outliers.

### Advantages

- Easy to understand.
- Shows data distribution clearly.
- Helps identify the shape of the data.

### Limitations

- Mainly used for numerical data.
- The appearance depends on the number of bins.
- It does not show exact individual values.



```python
import matplotlib.pyplot as plt

data = [10, 12, 15, 18, 20, 22, 25, 25, 28, 30]

plt.hist(data, bins=5, color="pink")

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")

plt.show()
```

### Interpretation

The Histogram shows how the values are distributed across different ranges.

In this example, most values are between 20 and 30.

The taller bars show the ranges where more values are present.

## 2. Bar Chart

### Definition

A Bar Chart is a graph used to compare different categories using rectangular bars.

The height or length of each bar represents the value of that category.

### Purpose

A Bar Chart helps us:

- Compare different categories.
- See which category has the highest or lowest value.
- Understand differences in data easily.

### When to Use

Use it when:

- Comparing categories.
- Showing counts or values.
- Finding the highest and lowest values.

### Advantages

- Simple and easy to understand.
- Good for comparing categories.
- Clearly shows differences between values.

### Limitations

- Not suitable for showing continuous data.
- Too many categories can make the chart difficult to read.



```python
import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "ML", "Stats"]
marks = [80, 75, 85, 70]

plt.bar(subjects, marks, color="red")

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Subject Marks")

plt.show()
```

### Interpretation

The Bar Chart compares the marks of different subjects.

The tallest bar shows the subject with the highest marks, while the shortest bar shows the subject with the lowest marks.

In this example, ML has the highest marks (85) and Statistics has the lowest marks (70).

## 3. Line Chart

### Definition

A Line Chart is a graph that uses points connected by lines to show changes in data.

### Purpose

A Line Chart helps us:

- See changes over time.
- Identify trends.
- Compare values across different points.
- Understand whether values are increasing or decreasing.

### When to Use

Use it when:

- Showing data over time.
- Tracking trends.
- Comparing changes between different points.

### Advantages

- Easy to see trends.
- Good for time-based data.
- Shows increases and decreases clearly.

### Limitations

- Not ideal for comparing many categories.
- Too many lines can make the chart difficult to understand.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 25, 22, 30, 35]

plt.plot(months, sales, marker="o", color="orange")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.show()
```

### Interpretation

The Line Chart shows how sales change from January to May.

Sales generally increase over time, with a small decrease in March.

The highest sales are recorded in May.

## 4. Scatter Plot

### Definition

A Scatter Plot is a graph that shows the relationship between two numerical variables.

It uses individual points to represent the values.

### Purpose

A Scatter Plot helps us:

- Understand the relationship between two variables.
- Identify positive or negative relationships.
- Find patterns in the data.
- Identify possible outliers.

### When to Use

Use it when:

- Comparing two numerical variables.
- Studying relationships between variables.
- Looking for patterns or trends.

### Advantages

- Easy to see relationships.
- Helps identify patterns and outliers.
- Useful for correlation analysis.

### Limitations

- Not suitable for categorical data.
- Too many points can make the graph difficult to read.

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5]
marks = [45, 50, 60, 65, 75]

plt.scatter(study_hours, marks, color="red")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

### Interpretation

The Scatter Plot shows the relationship between study hours and marks.

In this example, marks generally increase as study hours increase.

This shows a positive relationship between the two variables.

## 5. Pie Chart

### Definition

A Pie Chart is a circular graph used to show how different categories make up a whole.

Each section of the pie represents a category and its size shows its proportion.

### Purpose

A Pie Chart helps us:

- Understand parts of a whole.
- Compare proportions.
- See which category has a larger share.

### When to Use

Use it when:

- Showing percentages or proportions.
- Comparing a small number of categories.
- Showing how a total is divided.

### Advantages

- Easy to understand.
- Good for showing proportions.
- Simple and visually clear.

### Limitations

- Not suitable for many categories.
- Small differences can be difficult to compare.
- Not ideal for showing changes over time.

```python
import matplotlib.pyplot as plt

labels = ["Python", "SQL", "ML", "Statistics"]
values = [30, 25, 25, 20]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Learning Time")
plt.show()
```

### Interpretation

The Pie Chart shows how learning time is divided among different subjects.

Python has the highest share at **30%**, while Statistics has the lowest share at **20%**.

## 6. Density Plot

### Definition

A Density Plot is a graph used to show the distribution of numerical data.

It shows where the data values are more concentrated.

### Purpose

A Density Plot helps us:

- Understand the shape of the data.
- See where values are concentrated.
- Compare data distributions.
- Identify peaks in the data.

### When to Use

Use it when:

- You want to understand data distribution.
- You want a smoother view than a Histogram.
- You want to compare distributions.

### Advantages

- Easy to see the shape of data.
- Shows the concentration of values clearly.
- Useful for comparing distributions.

### Limitations

- Mainly used for numerical data.
- The shape can change depending on the smoothing method.
- It may be harder to understand for beginners.

```python
import matplotlib.pyplot as plt

data = [10, 12, 15, 18, 20, 22, 25, 25, 28, 30]

plt.hist(data, bins=5, density=True, color="red", alpha=0.5)

plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Density Plot")

plt.show()
```

### Interpretation

The Density Plot shows where the data values are concentrated.

A higher part of the curve means more values are concentrated in that range.

## 7. Box Plot

### Definition

A Box Plot is a graph used to show the spread and distribution of numerical data.

It shows the minimum, maximum, median, and quartiles of the data.

### Purpose

A Box Plot helps us:

- Understand the spread of data.
- Find the middle value (median).
- Identify possible outliers.
- Compare different groups.

### When to Use

Use it when:

- Checking the distribution of numerical data.
- Finding outliers.
- Comparing multiple groups.

### Advantages

- Easy to identify outliers.
- Shows the spread of data clearly.
- Useful for comparing groups.

### Limitations

- Does not show every individual value.
- Can be difficult to understand for beginners.
- Gives less detail about the exact distribution.

```python
import matplotlib.pyplot as plt

data = [10, 12, 15, 18, 20, 22, 25, 25, 28, 30, 45]

plt.boxplot(data)

plt.ylabel("Values")
plt.title("Box Plot")

plt.show()
```

### Interpretation

The Box Plot shows the spread of the data and its median.

The point outside the main box and whiskers may represent an outlier.

In this example, **45 may be an outlier** because it is much higher than the other values.

## 8. Violin Plot

### Definition

A Violin Plot is a graph used to show the distribution and spread of numerical data.

It combines the features of a Box Plot and a Density Plot.

### Purpose

A Violin Plot helps us:

- Understand the shape of the data.
- See where values are concentrated.
- Understand the spread of data.
- Compare different groups.

### When to Use

Use it when:

- Comparing the distribution of different groups.
- Understanding where data is concentrated.
- You need more detail than a Box Plot.

### Advantages

- Shows the distribution clearly.
- Shows the spread and concentration of data.
- Useful for comparing groups.

### Limitations

- Can be difficult to understand for beginners.
- Not suitable for very small datasets.
- More complex than a Box Plot.

```python
import matplotlib.pyplot as plt

data = [10, 12, 15, 18, 20, 22, 25, 25, 28, 30]

plt.violinplot(data)

plt.ylabel("Values")
plt.title("Violin Plot")

plt.show()
```

### Interpretation

The Violin Plot shows how the data is distributed.

The wider part shows where more values are concentrated, while the narrow part shows fewer values.

## 9. Pair Plot

### Definition

A Pair Plot is a collection of graphs that shows the relationships between multiple numerical features in a dataset.

It combines scatter plots and distribution plots in one view.

### Purpose

A Pair Plot helps us:

- Understand relationships between multiple features.
- Find patterns and trends.
- Compare different features.
- Identify possible outliers.

### When to Use

Use it when:

- You have multiple numerical features.
- You want to quickly explore relationships in a dataset.
- You want to find patterns before building a model.

### Advantages

- Shows many relationships at once.
- Easy to compare multiple features.
- Useful for initial data analysis.

### Limitations

- Can become difficult to read with many features.
- Mainly useful for numerical features.
- Can take more time to display for large datasets.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("iris")

sns.pairplot(data)
plt.show()
```

### Interpretation

The Pair Plot shows the relationships between different features.

Scatter plots show relationships between two features, while the plots on the diagonal show the distribution of each feature.

## 10. Heatmap

### Definition

A Heatmap is a graph that uses different colors to show values in a table or matrix.

It makes it easy to see high, low, and related values.

### Purpose

A Heatmap helps us:

- Understand patterns quickly.
- Compare values easily.
- Find strong and weak relationships.
- Visualize a Correlation Matrix.

### When to Use

Use it when:

- Working with a Correlation Matrix.
- Comparing many values.
- Looking for patterns in data.

### Advantages

- Easy to understand visually.
- Shows patterns clearly.
- Useful for correlation analysis.

### Limitations

- Too many values can make it difficult to read.
- Colors need to be interpreted correctly.
- It may not show exact values clearly without labels.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("iris")

sns.heatmap(data.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation

The Heatmap shows the correlation between different features.

Values closer to **+1** show a strong positive relationship, while values closer to **-1** show a strong negative relationship.

Values close to **0** show a weak or no linear relationship.

##  Advantages
