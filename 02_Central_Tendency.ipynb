# Module 2 - Measures of Central Tendency

Measures of Central Tendency are used to find a central or typical value from a group of data.

The main measures covered in this module are:

- Mean
- Median
- Mode
- Weighted Mean
- Geometric Mean
- Harmonic Mean

## 1. Mean
### Definition

- Mean represents the average value of a set of numbers.
- It is calculated by adding all the values and dividing the total by the number of values.

### Formula

$$
Mean = \frac{\text{Sum of all values}}{\text{Number of values}}
$$

### Numerical Example

Consider the numbers:

4, 8, 6, 5, 3, 7

First, add all the values:

4 + 8 + 6 + 5 + 3 + 7 = 33

There are 6 values.

Mean = 33 / 6 = 5.5

Therefore, the mean is 5.5.

```python
numbers = [4, 8, 6, 5, 3, 7]

total = sum(numbers)
count = len(numbers)

mean = total / count

print("Numbers:", numbers)
print("Mean:", mean)
```

### Business Example

A cafe can calculate the average number of cups sold each day to understand its normal daily sales.

### AI/ML Example

In AI/ML, the mean can be used to understand the average value of numerical data such as age, salary, or price.

### Interpretation

The mean is 5.5, which represents the average value of the given numbers.

## 2. Median
### Definition
* Median is the middle value of a dataset after arranging the values in order.

* If there are two middle values, their average is taken as the median.

### Formula

For an odd number of values:

$$
Median = \text{Middle value}
$$

For an even number of values:

$$
Median = \frac{\text{Sum of the two middle values}}{2}
$$

### Numerical Example

Consider the numbers:

3, 7, 5, 9, 2

Arrange them in ascending order:

2, 3, 5, 7, 9

The middle value is 5.

Therefore, the median is 5.

```python
numbers = [3, 7, 5, 9, 2]

numbers.sort()
n = len(numbers)

if n % 2 == 1:
    median = numbers[n // 2]
else:
    median = (numbers[n // 2 - 1] + numbers[n // 2]) / 2

print("Sorted Numbers:", numbers)
print("Median:", median)
```

### Business Example

A company can use the median salary to understand the typical salary of its employees when a few employees have very high salaries.

### AI/ML Example

Median can be useful when a dataset contains extreme values or outliers.

### Interpretation

The median is 5 because it is the middle value after arranging the numbers in order.

## 3. Mode

### Definition

- Mode is the value that occurs most frequently in a dataset.
- It helps us identify the most common value or category.

### Formula
$$\textit{Mode} = \text{The value that occurs the most number of times}$$

### Numerical Example

Consider the numbers:

2, 3, 3, 5, 7, 3, 8

The number 3 appears three times.

Therefore, the mode is 3.

```python
from collections import Counter

numbers = [2, 3, 3, 5, 7, 3, 8]

frequency = Counter(numbers)
mode = frequency.most_common(1)[0][0]

print("Numbers:", numbers)
print("Mode:", mode)
```

### Business Example

A shoe store can use the mode to find the shoe size that is sold most often and keep more of that size in stock.

### AI/ML Example

Mode can be used to find the most common category when working with categorical data.

### Interpretation

The mode is 3 because it occurs more often than the other values.

## 4. Weighted Mean
### Definition
* Weighted Mean is an average where different values have different levels of importance.

* A value with a higher weight has a greater effect on the final result.

### Formula

$$
Weighted\ Mean =
\frac{\sum(x \times w)}{\sum w}
$$

Where:

- $x$ = value
- $w$ = weight

### Numerical Example

A student gets the following scores:

- Exam 1 = 80, weight = 20%
- Exam 2 = 90, weight = 30%
- Exam 3 = 70, weight = 50%

Weighted Mean:

(80 × 0.20) + (90 × 0.30) + (70 × 0.50)

= 16 + 27 + 35

= 78

Therefore, the weighted mean is 78.

```python
scores = [80, 90, 70]
weights = [0.20, 0.30, 0.50]

weighted_mean = sum(
    score * weight for score, weight in zip(scores, weights)
) / sum(weights)

print("Weighted Mean:", weighted_mean)
```

### Business Example

A company can calculate an employee performance score by giving different weights to different performance factors.

### AI/ML Example

Weighted averages can be used when some values or predictions need to have more importance than others.

### Interpretation

The weighted mean is 78. The third exam has the highest weight, so it has the greatest effect on the final result.

## 5. Geometric Mean
### Definition

* Geometric Mean is an average calculated by multiplying the values together and taking the appropriate root.

* It is commonly used for growth rates and values that change over time.

### Formula

$$
Geometric\ Mean =
\sqrt[n]{x_1 \times x_2 \times \cdots \times x_n}
$$

Here, $n$ represents the number of values.

### Numerical Example

Consider the numbers:

2, 4, 8

First, multiply the values:

2 × 4 × 8 = 64

There are 3 values, so we take the cube root:

$$
\sqrt[3]{64} = 4
$$

Therefore, the geometric mean is 4.

### Business Example

Geometric Mean can be used to understand the average growth of an investment over several years.

### AI/ML Example

Geometric Mean can be used when working with growth rates, ratios, or values that change multiplicatively.

### Interpretation

The geometric mean is 4. It gives an average that is suitable for values that are multiplied or compounded over time.

## 6. Harmonic Mean
### Definition

*  Harmonic Mean is an average that is mainly used for rates and ratios.

*  It gives more importance to smaller values compared with the arithmetic mean.

### Formula

$$
Harmonic\ Mean =
\frac{n}{\sum \frac{1}{x_i}}
$$

Where:

- $n$ = number of values
- $x_i$ = individual values

### Numerical Example

Consider the numbers:

2, 4, 8

First, calculate the reciprocals:

1/2 + 1/4 + 1/8

= 0.5 + 0.25 + 0.125

= 0.875

There are 3 values.

Therefore:

Harmonic Mean = 3 / 0.875

≈ 3.43

So, the harmonic mean is approximately 3.43.

```python
numbers = [2, 4, 8]

harmonic_mean = len(numbers) / sum(1 / number for number in numbers)

print("Numbers:", numbers)
print("Harmonic Mean:", round(harmonic_mean, 2))
```

### Business Example

Harmonic Mean can be used to calculate an average rate, such as average speed when travelling the same distance at different speeds.

### AI/ML Example

Harmonic Mean is used in AI/ML through the F1-score, which combines precision and recall.

### Interpretation

The harmonic mean is approximately 3.43. It is useful when working with rates and ratios.

## Quick Comparison

| Measure | What it tells us | When it is useful |
|---|---|---|
| Mean | Gives the average value of all the data. | Useful for finding a general average. |
| Median | Gives the middle value after arranging the data. | Useful when the data has very high or very low values. |
| Mode | Gives the value that appears most often. | Useful for finding the most common value or category. |
| Weighted Mean | Gives an average where some values have more importance than others. | Useful when different values have different weights. |
| Geometric Mean | Gives an average for values that change by multiplication or growth. | Useful for growth rates and investment returns. |
| Harmonic Mean | Gives an average that is useful for rates and ratios. | Useful for speed, rates, and other ratio-based data. |

## Conclusion

- Mean, Median, and Mode are common ways to find a central value in data.
- Weighted Mean is useful when different values have different importance.
- Geometric Mean is useful for growth and compounded values.
- Harmonic Mean is useful for rates and ratios.
- These measures help us summarize data and understand it more easily.
- They are also useful in business analysis and AI/ML applications.
