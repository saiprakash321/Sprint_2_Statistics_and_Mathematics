# Introduction to Hypothesis Testing

### Definition

Hypothesis Testing is a statistical method used to make decisions about a population using sample data.

It helps us check whether there is enough evidence to support a claim or whether the observed result could have happened by chance.

### Why Is It Used?

Hypothesis Testing is used to:

- Compare two groups
- Check whether a change is significant
- Test assumptions about data
- Support decisions using statistical evidence
- Analyze results in Data Science and Machine Learning

### Simple Example

Suppose a company claims that a new website design increases the average number of sales.

We can collect sample data and use Hypothesis Testing to check whether the increase is statistically significant.

### Basic Process

1. Define the Null and Alternative Hypotheses.
2. Choose a Significance Level.
3. Collect and analyze sample data.
4. Calculate the p-value.
5. Make a decision about the Null Hypothesis.

## 1. Null Hypothesis

### Definition

The Null Hypothesis (H₀) is the assumption that there is no significant difference, relationship, or effect.

It is the starting point of a hypothesis test.

### Simple Example

Suppose a company wants to check whether a new training program improves employee performance.

The Null Hypothesis would be:

**H₀: The new training program does not improve employee performance.**

### Where It Is Used

The Null Hypothesis is used in:

- Statistical Testing
- A/B Testing
- Medical Research
- Business Analysis
- Data Science
- Machine Learning

### Important Point

If the evidence from the data is strong enough, we reject the Null Hypothesis.

If the evidence is not strong enough, we fail to reject the Null Hypothesis.

## 2. Alternative Hypothesis

### Definition

The Alternative Hypothesis (H₁) is the statement that there is a significant difference, relationship, or effect.

It is considered when the data provides enough evidence against the Null Hypothesis.

### Simple Example

Suppose a company wants to check whether a new training program improves employee performance.

The Alternative Hypothesis would be:

**H₁: The new training program improves employee performance.**

### Where It Is Used

The Alternative Hypothesis is used in:

- Statistical Testing
- A/B Testing
- Medical Research
- Business Analysis
- Data Science
- Machine Learning

### Important Point

The Alternative Hypothesis represents the result we are trying to find evidence for.

If the evidence is strong enough, we reject the Null Hypothesis in favor of the Alternative Hypothesis.

## 3. p-value

### Definition

The p-value tells us how strong the evidence is against the Null Hypothesis.

A small p-value means there is strong evidence against the Null Hypothesis.

### Simple Rule

- **p-value < 0.05** → Reject the Null Hypothesis
- **p-value ≥ 0.05** → Fail to reject the Null Hypothesis

### Simple Example

If a test gives a p-value of 0.02, it is less than 0.05.

Therefore, we reject the Null Hypothesis.

### Where It Is Used

p-value is commonly used in:

- Statistical Testing
- A/B Testing
- Data Science
- Research
- Machine Learning

### Python Example

Python can be used to calculate the p-value using statistical tests.

```python
from scipy.stats import ttest_1samp

data = [52, 54, 51, 53, 55, 52, 54]

t_stat, p_value = ttest_1samp(data, 50)

print("p-value:", p_value)
```

### Explanation

The p-value helps us decide whether the observed result provides enough evidence to reject the Null Hypothesis.

## 4. Confidence Interval

### Definition

A Confidence Interval is a range of values that is used to estimate a population parameter from sample data.

It gives us an estimated range instead of using only one value.

### Simple Example

If the average marks of a sample are 70 and the 95% confidence interval is 67 to 73, the estimated range is 67 to 73.

### Where It Is Used

Confidence Intervals are commonly used in:

- Statistics
- Research
- Surveys
- Data Analysis
- Machine Learning

### Important Point

A 95% confidence interval is commonly used in statistical analysis.

### Python Example

Python can be used to calculate a confidence interval for sample data.

```python
import numpy as np
from scipy.stats import t

data = np.array([65, 70, 72, 68, 75, 71, 69])

mean = np.mean(data)
std = np.std(data, ddof=1)
sem = std / np.sqrt(len(data))

confidence = t.interval(
    0.95,
    len(data) - 1,
    loc=mean,
    scale=sem
)

print("Mean:", mean)
print("95% Confidence Interval:", confidence)
```

### Explanation

The code calculates the sample mean and a 95% confidence interval.

The interval gives a range of plausible values for the population mean.

## 5. Significance Level

### Definition

The Significance Level is the threshold used to decide whether a statistical result is significant.

It is usually represented by **α (alpha)**.

A commonly used significance level is **0.05**.

### Simple Example

If α = 0.05:

- p-value < 0.05 → Reject H₀
- p-value ≥ 0.05 → Fail to reject H₀

### Where It Is Used

Significance Level is used in:

- Hypothesis Testing
- Scientific Research
- A/B Testing
- Data Analysis
- Medical Studies

### Important Point

The significance level is chosen before performing the hypothesis test.

### Python Example

We can compare a p-value with the significance level using Python.

```python
p_value = 0.03
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to reject the Null Hypothesis")
```

### Explanation

Here, the p-value is 0.03 and the significance level is 0.05.

Since 0.03 is less than 0.05, we reject the Null Hypothesis.

## 6. Type I Error

### Definition

Type I Error occurs when we reject the Null Hypothesis even though it is actually true.

It is also called a **False Positive**.

### Simple Example

Suppose a medical test says that a person has a disease when the person actually does not have the disease.

This is a Type I Error.

### Where It Is Used

Type I Error is important in:

- Medical Testing
- Hypothesis Testing
- A/B Testing
- Research
- Data Science

### Important Point

Type I Error is related to the significance level (α).

### Easy Way to Remember

**Type I Error → Reject a true Null Hypothesis.**

## 7. Type II Error

### Definition

Type II Error occurs when we fail to reject the Null Hypothesis even though it is actually false.

It is also called a **False Negative**.

### Simple Example

Suppose a medical test says that a person does not have a disease when the person actually has the disease.

This is a Type II Error.

### Where It Is Used

Type II Error is important in:

- Medical Testing
- Hypothesis Testing
- Research
- A/B Testing
- Data Science

### Important Point

Type II Error means that a real effect or difference was not detected.

### Easy Way to Remember

**Type II Error → Fail to reject a false Null Hypothesis.**

## 8. One-tailed Test

### Definition

A One-tailed Test is a hypothesis test that checks for a difference or effect in one specific direction.

It can test whether a value is significantly greater than or significantly less than a given value.

### Simple Example

A company wants to check whether a new training program increases employee performance.

The test is interested only in an **increase**.

### Where It Is Used

One-tailed tests are used when:

- The direction of the effect is known
- Testing an increase
- Testing a decrease
- A/B Testing
- Business Experiments

### Important Point

A One-tailed Test has one rejection region.

### Python Example

Python can perform a one-tailed test using statistical libraries.

```python
from scipy.stats import ttest_1samp

data = [52, 54, 51, 53, 55, 52, 54]

t_stat, p_value = ttest_1samp(data, 50)

one_tail_p = p_value / 2

print("One-tailed p-value:", one_tail_p)
```

### Explanation

A one-tailed test checks for an effect in one specific direction, such as only an increase or only a decrease.

## 9. Two-tailed Test

### Definition

A Two-tailed Test checks whether there is a significant difference in either direction.

It can detect both an increase and a decrease.

### Simple Example

Suppose we want to check whether a new teaching method changes student performance.

We are interested in both:

- Performance increasing
- Performance decreasing

### Where It Is Used

Two-tailed tests are commonly used in:

- Research
- Data Analysis
- A/B Testing
- Medical Studies
- Machine Learning

### Important Point

A Two-tailed Test has two rejection regions, one on each side of the distribution.

### Python Example


```python
from scipy.stats import ttest_1samp

data = [52, 54, 51, 53, 55, 52, 54]

t_stat, p_value = ttest_1samp(data, 50)

print("Two-tailed p-value:", p_value)
```

### Explanation

The test checks whether the sample mean is significantly different from the given value in either direction.

This is the default behavior of `ttest_1samp()`.

## Advantages

- Helps make decisions using data.
- Helps test claims and assumptions.
- Useful for comparing groups.
- Helps identify statistically significant results.
- Widely used in Data Science and Machine Learning.

##  Limitations

- Results depend on the quality of the sample data.
- A small sample may give unreliable results.
- p-value does not tell us how important an effect is.
- Statistical significance does not always mean practical importance.
- Choosing the correct test can sometimes be difficult.
