# Module 1 - Introduction to Statistics

## 1. What is Statistics?

- Statistics is a way of working with data to understand what is happening and make useful decisions.
- It involves collecting, organizing, analyzing, and interpreting data.
- In simple words, statistics helps us understand what the data is telling us.

Statistics is useful in many areas such as healthcare, business, finance, education, and AI/ML.

### Example

Imagine a hospital records the ages of patients who visit during a day.

For example, the ages are 20, 25, 30, 35, and 40.

By calculating the average age, the hospital can get a quick idea about the age group of the patients it serves.


### Python Implementation

Here, we will calculate the average age of a small group of patients.

```python
ages = [20, 25, 30, 35, 40]

total_age = sum(ages)
number_of_patients = len(ages)

average_age = total_age / number_of_patients

print("Patient Ages:", ages)
print("Average Age:", average_age)
```

### Explanation

The program stores the ages of five patients in a list. It calculates the total age and divides it by the number of patients to find the average.

- `sum(ages)` calculates the total age.
- `len(ages)` counts the number of patients.

The average age is 30 years.

### Statistical Concept Demonstrated

This example demonstrates how an average can be used to summarize a group of data.

### Output Interpretation

The output shows an average age of 30 years. This gives us a simple summary of the ages in the group.



## 2. Why Statistics is Important in AI/ML?

Statistics is important in AI/ML because Machine Learning models learn from data. Before building a model, we need to understand the data properly.

Statistics helps us to:

- Understand how the data is distributed.
- Find missing values and unusual data.
- Identify useful features.
- Understand relationships between variables.
- Evaluate model performance.
- Make better predictions and decisions.

### Example

Suppose a bank wants to predict whether a customer will repay a loan.

Statistics can help the bank understand:

- Customer income
- Age
- Loan amount
- Previous repayment history

After understanding the data, a Machine Learning model can use these patterns to make predictions.

### AI/ML Application

Statistics is mainly used to understand and prepare data before building Machine Learning models. It also helps in evaluating the results produced by the model.

## 3. Statistics vs Machine Learning

Statistics and Machine Learning both use data, but they are mainly used for different purposes.
| No. | Statistics | Machine Learning |
|---|---|---|
| 1 | Mainly used to understand data. | Mainly used to learn from data. |
| 2 | Helps to find patterns and relationships. | Uses those patterns to make predictions. |
| 3 | Helps us understand why something is happening in the data. | Helps us predict what can happen next. |
| 4 | Uses formulas and statistical methods. | Uses algorithms and models. |

### Example

Consider a bank that has customer loan data.

- **Statistics:** The bank can analyze the average income of customers and study the relationship between income and loan repayment.
- **Machine Learning:** The bank can use previous customer data to predict whether a new customer is likely to repay a loan.

### In Simple Words

**Statistics → Understand the data**

**Machine Learning → Learn from the data and make predictions**


## 4. Descriptive Statistics

- Descriptive statistics is used to summarize and understand a set of data.
- It gives us a quick idea about the data without looking at every value separately.

Some common measures are:

- Mean
- Median
- Mode
- Minimum
- Maximum
- Standard Deviation

### Example

Suppose a teacher has the following marks of five students:

80, 75, 90, 95, 88

Using descriptive statistics, the teacher can find the average mark, highest mark, lowest mark, and how much the marks vary.

### Python Implementation

We can use Pandas to get a summary of the marks.

```python
import pandas as pd

marks = pd.Series([80, 75, 90, 95, 88])

print(marks.describe())
```

### Explanation

The `describe()` function gives a quick summary of the marks.

- `count` shows the number of values.
- `mean` shows the average.
- `std` shows how much the marks vary.
- `min` shows the lowest mark.
- `max` shows the highest mark.
- `50%` represents the median.

### Output Interpretation

The average mark is 85.6, while the lowest mark is 75 and the highest mark is 95.

This gives us a quick understanding of how the students performed.

### AI/ML Application

Descriptive statistics is useful in AI/ML for understanding and summarizing datasets before building a model.

## 5. Inferential Statistics

- Inferential statistics is used to make conclusions about a larger group using data from a smaller group.
- Instead of studying the entire population, we study a sample and use its results to understand the population.
- It is useful when collecting information from everyone is difficult or time-consuming.

### Example

A company has 20,000 customers but wants to know whether customers are satisfied with its service.

Instead of asking all 20,000 customers, the company surveys 500 customers and uses their responses to estimate the satisfaction level of all customers.

### Simple Difference

- Descriptive Statistics → Summarizes the data we have.
- Inferential Statistics → Uses sample data to make conclusions about a larger population.

### AI/ML Application

Inferential statistics can help in AI/ML when we use a sample of data to understand a larger population and make decisions based on the results.

## 6. Population

- Population means the complete group of people, objects, or data that we want to study.
- It includes every member of the group that is relevant to the study.

### Example

If we want to study the performance of students in a college:

- All students in the college are the population.
- We can study their marks, attendance, or other details.

### Another Example

If a company wants to know the satisfaction of all its customers, then all its customers together form the population.

### AI/ML Application

In AI/ML, a population can represent the complete set of people, products, transactions, or other data that we want to understand or make predictions about.

## 7. Sample

- A sample is a smaller group selected from a population for study.
- It is used when studying the entire population is difficult or time-consuming.
- The sample should represent the population as closely as possible.

### Example

A college has 5,000 students, but we want to study their average marks.

Instead of collecting marks from all 5,000 students, we select 200 students and study their marks.

- Population → All 5,000 students
- Sample → Selected 200 students

### AI/ML Application

In AI/ML, samples are often used to analyze a smaller portion of a large dataset and understand the overall data.

## 8. Census

- A census is a method of collecting information from every member of a population.
- It means the entire population is studied instead of selecting only a sample.
- A census can provide detailed and complete information about the population.

### Example

Suppose a college has 5,000 students and wants to know their average marks.

If the college collects the marks of all 5,000 students, it is called a census.

### Simple Difference

- Census → Studies the entire population.
- Sample → Studies only a part of the population.

## 9. Sampling

- Sampling is the process of selecting a smaller group from a larger population for study.
- It helps us collect information without studying every member of the population.
- A good sample should represent the population as closely as possible.

### Example

A company has 10,000 customers and wants to know how satisfied they are.

Instead of asking all 10,000 customers, the company selects 500 customers and collects their feedback.

- Population → 10,000 customers
- Sample → 500 selected customers
- Sampling → The process of selecting those 500 customers

### AI/ML Application

Sampling is useful in AI/ML when a dataset is very large. A suitable sample can be used to understand the data and perform analysis more efficiently.

### Sample vs Sampling

| Sample | Sampling |
|---|---|
| A sample is a smaller group selected from a population. | Sampling is the process of selecting that smaller group. |
| It is the actual group being studied. | It is the method used to choose the group. |
| Example: 500 customers selected from 10,000 customers. | Example: The process of selecting those 500 customers. |

## 10. Variables

- A variable is a characteristic or value that can change from one observation to another.
- Variables are used to describe different properties of the data.
- The value of a variable can be different for different people, objects, or records.

### Examples

Some common variables are:

- Age
- Height
- Salary
- Gender
- City
- Marks

### Example

Consider the details of employees in a company.

- Age can be different for each employee.
- Salary can be different for each employee.
- Department can also be different.

Therefore, age, salary, and department are examples of variables.

### AI/ML Application

In AI/ML, variables are used to represent information about the data. They can be used as inputs to understand patterns and make predictions.

## 11. Features

- Features are the input values or characteristics used by a Machine Learning model.
- They provide information that helps the model understand the data and make predictions.
- Features are also called input variables or independent variables.

### Example

Suppose we want to predict the price of a house.

Some features can be:

- Area of the house
- Number of bedrooms
- Location
- Age of the house

The **house price** is the target that we want to predict.

### AI/ML Application

Features are given as inputs to a Machine Learning model. The model learns from these features to make predictions.

## 12. Observation

- An observation is a single record or row in a dataset.
- It represents one person, object, event, or case being studied.
- Each observation contains the values of different variables.

### Example

Consider a dataset containing student details.

| Name | Age | Marks |
|---|---:|---:|
| A | 20 | 80 |
| B | 21 | 90 |
| C | 20 | 75 |

Here, each row represents **one observation**.

### AI/ML Application

In AI/ML, each observation represents one data record that a model can use to learn patterns and make predictions.

## 13. Dataset

- A dataset is a collection of related data arranged in rows and columns.
- Each row represents an observation.
- Each column represents a variable or feature.
- Datasets are commonly used for analysis and Machine Learning.

### Example

Consider a dataset containing student information.

| Name | Age | Marks |
|---|---:|---:|
| A | 20 | 80 |
| B | 21 | 90 |
| C | 20 | 75 |

Here, the complete table is a dataset.

### AI/ML Application

In AI/ML, datasets are used to train, test, and evaluate Machine Learning models.

## 14. Quantitative Data

- Quantitative data contains numerical values that can be measured or counted.
- It represents information that can be expressed using numbers.
- It can be divided into continuous and discrete data.

### Examples

- Age
- Height
- Weight
- Salary
- Number of students

### Example

The salaries of three employees are:

30000, 40000, and 50000.

These are quantitative data because they are numerical values.

### AI/ML Application

Quantitative data is commonly used as input for Machine Learning models to find patterns and make predictions.

## 15. Qualitative Data

- Qualitative data represents categories or labels rather than numerical measurements.
- It describes qualities or characteristics of people, objects, or records.
- It is usually expressed using words or categories.

### Examples

- Gender
- Color
- City
- Blood Group
- Department

### Example

The colors of three cars are:

Red, Blue, and Green.

These are qualitative data because they represent categories.

### AI/ML Application

Qualitative data can be converted into a suitable numerical form and then used by Machine Learning models.

### Quantitative Data vs Qualitative Data

| Quantitative Data | Qualitative Data |
|---|---|
| Represents numerical values. | Represents categories or labels. |
| Can be measured or counted. | Describes qualities or characteristics. |
| Example: Age, Salary, Height | Example: Color, City, Blood Group |

## 16. Continuous Data

- Continuous data can take any value within a given range.
- It is usually obtained by measuring something.
- It can have decimal values.

### Examples

- Height
- Weight
- Temperature
- Time

### Example

A person's height can be:

170.5 cm

The value can also be 170.51 cm or 170.55 cm, depending on the accuracy of measurement.

### AI/ML Application

Continuous data is commonly used in Machine Learning for tasks such as predicting temperature, weight, house prices, and other measurable values.

## 17. Discrete Data

- Discrete data consists of values that can be counted.
- It usually contains whole numbers.
- It cannot normally take values between two whole numbers.

### Examples

- Number of students
- Number of cars
- Number of books
- Number of employees

### Example

A classroom has 35 students.

The number of students can be 35 or 36, but it cannot be 35.5.

### AI/ML Application

Discrete data can be used in Machine Learning to analyze counts and make predictions based on numerical patterns.

### Continuous Data vs Discrete Data

| Continuous Data | Discrete Data |
|---|---|
| Can have decimal values. | Usually contains whole numbers. |
| Usually obtained by measuring. | Usually obtained by counting. |
| Example: Height, Weight, Temperature | Example: Students, Cars, Books |

## 18. Structured Data

- Structured data is organized in a fixed format, usually in rows and columns.
- It is easy to store, search, and analyze.
- It is commonly found in databases and spreadsheets.

### Examples

- SQL databases
- Excel files
- CSV files
- Employee records

### Example

| Name | Age | Salary |
|---|---:|---:|
| A | 22 | 30000 |
| B | 25 | 40000 |

This is structured data because the information is organized into rows and columns.

### AI/ML Application

Structured data is commonly used to train Machine Learning models because it is already organized and easier to process.

## 19. Unstructured Data

- Unstructured data does not have a fixed format like rows and columns.
- It can contain different types of information and is usually more difficult to analyze.
- It can include text, images, audio, and video.

### Examples

- Images
- Videos
- Audio
- Emails
- Social media posts
- PDF documents

### Example

A customer sends an email describing a problem with a product.

The email is unstructured data because it does not follow a fixed table format.

### AI/ML Application

AI/ML techniques can be used to process unstructured data such as images, text, audio, and videos to find useful information and patterns.

### Structured Data vs Unstructured Data

| Structured Data | Unstructured Data |
|---|---|
| Has a fixed format. | Does not have a fixed format. |
| Usually arranged in rows and columns. | Can be text, images, audio, or video. |
| Example: SQL table, Excel, CSV | Example: Images, Videos, Emails |

## Advantages of Statistics

- Helps us understand and summarize data.
- Helps identify patterns and relationships in data.
- Supports better decision-making.
- Helps compare different groups or datasets.
- Provides a foundation for AI and Machine Learning.
- Helps prepare and understand data before building ML models.

## Limitations of Statistics

- Results depend on the quality of the data.
- Incorrect or incomplete data can lead to wrong conclusions.
- Statistical results may not always explain the reason behind a pattern.
- A sample may not perfectly represent the entire population.
- Statistical methods need to be selected carefully for different types of data.
